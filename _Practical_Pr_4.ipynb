# PART A: VECTOR & MATRIX FUNDAMENTALS

```python
import numpy as np
import pandas as pd 


Data = {
    "Mathematics": [80, 65, 90, 55, 75],
    "Statistics": [75, 70, 85, 60, 80],
    "Python": [90, 60, 95, 65, 70]
}

DF = pd.DataFrame(Data, index=["S1", "S2", "S3", "S4", "S5"])

print(DF)
```

## 1. Represent Each Student's Subject Scores as a Vector.

```python
X = DF.to_numpy()
print(X)
```

```python
print("S1 Vector:", X[0])
print("S2 Vector:", X[1])
print("S3 Vector:", X[2])
print("S4 Vector:", X[3])
print("S5 Vector:", X[4])
```

## 2. Compute Norm-1 and Norm-2 of Vectors

```python
L1 = np.linalg.norm(X[0], ord=1)

print("Norm-1:", L1)
```

```python
L2 = np.linalg.norm(X[0], ord=2)

print("Norm-2:", round(L2, 2))
```

## 2. Calculate Dot Product and Angle Between Two Students' Score Vectors

```python
A = X[0]
B = X[1]

dot_product = np.dot(A, B)

print("Dot Product:", dot_product)
```

## 3. Calculate Cross Product (for 3D Students)

```python
cross_product = np.cross(A, B)

print("Cross Product:", cross_product)
```

# PART B: MATRIX OPERATIONS

## 4. Form a Matrix of Students × Subjects. Perform:

```python
M = DF.to_numpy()

print("Student Marks Matrix:")
print(M)
```

```python
Addition = M + M

print("Matrix Addition:")
print(Addition)
```

```python
Multiplication = M @ M.T

print("Matrix Multiplication:")
print(Multiplication)
```

## Transpose of Matrix

```python
Transpose = M.T

print("Transpose Matrix:")
print(Transpose)
```

```python
Square_Matrix = M.T @ M

Inverse = np.linalg.inv(Square_Matrix)

print("Inverse Matrix:")
print(Inverse)
```

## Determinant

```python
Determinant = np.linalg.det(Square_Matrix)

print("Determinant:")
print(round(Determinant, 2))
```

# PART C: LINEAR TRANSFORMATIONS & GEOMETRY

## 5. Explain Line, Plane, and Hyperplane with Respect to Your Dataset Dimensions.

```python

A_2D = np.array([[2, 0],
                 [0, 2]])

Vector_2D = np.array([3, 4])

Transformation_2D = A_2D @ Vector_2D

print("Original 2D Vector:", Vector_2D)
print("Transformation 2D Vector:", Transformation_2D)

print("\nLine: 1 Dimension")
print("Plane: 2 Dimensions")
print("Hyperplane: n - 1 Dimensions")
```

```python

A_3D = np.array([[2, 0, 0],
                 [0, 2, 0],
                 [0, 0, 2]])

Vector_3D = np.array([80, 75, 90])

Transformation_3D = A_3D @ Vector_3D

print("Original 3D Vector:", Vector_3D)
print("Transformation 3D Vector:", Transformation_3D)

print("\nHyperplane in 3D is a 2D Plane")
```

## Part D:Eigenvalues & Decomposition
## 7. Compute Eigenvalues and Eigenvectors of Covariance Matrix.

```python

Cov_Matrix = np.cov(X, rowvar=False)

print("Covariance Matrix:\n", Cov_Matrix)
```

```python

Eigenvalues, Eigenvectors = np.linalg.eigh(Cov_Matrix)

print("Eigenvalues:\n", Eigenvalues)

print("\nEigenvectors:\n", Eigenvectors)
```
